```python
from scipy.io import loadmat

def get_nstimuli(file_path):
    """Loads only the subject field to save memory."""
    # variable_names restricts what is read from the disk
    data = loadmat(file_path, variable_names=['bins'], simplify_cells=True)
    return data['bins'].shape[0]

nstim = {f: get_nstimuli(os.path.join(data_dir, f)) for f in os.listdir(data_dir) if f.endswith('.mat')}
nstim
```
