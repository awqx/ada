# Longitudional, multi-region visual cortex recordings given texture stimuli

Data can be found on Zenodo. 

## Recording details

Lee et al. included 8 monkeys, with experimental details in Table 1 (copied below). Recording ages marked with `[a]` were used in the comparison in Figure 4D. Table 1 reports the number of visually responsive sites (more on this later). 

Recordings were made from a pair of 96 electrode Utah arrays per monkey (i.e., a total of 192 channels per subject). Lee et al. implanted arrays twice in M6, so the subject is repeated in the table. FC stands for "foveal confluence" and indicates measurements for where the specific area could not be determined with certainty. 

Two monkeys (M4 and M5) only underwent behavioral experiments. One monkey (M8) only underwent passive fixation recording. Two monkeys (M6 and M7) were tested cross-sectionally (i.e., only underwent passive fixation recordings at around 4 years old and not previously exposed to behavioral tests earlier in development).
